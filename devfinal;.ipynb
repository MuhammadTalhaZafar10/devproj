```python
# ============================================================
# 0. INSTALL / IMPORT LIBRARIES
# ============================================================

!pip -q install kagglehub lightgbm

import os
import glob
import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub
import lightgbm as lgb

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")

np.random.seed(42)

print("=" * 80)
print("BUILDING DATA GENOME PROJECT 2")
print("REAL DATASET - 100,000 ROWS - 35+ FEATURES")
print("=" * 80)


# ============================================================
# 1. DOWNLOAD REAL BDG2 DATASET FROM KAGGLE
# ============================================================

print("\n=== STEP 1: DOWNLOADING BDG2 FROM KAGGLE ===")

path = kagglehub.dataset_download(
    "claytonmiller/buildingdatagenomeproject2"
)

print("\nPath to dataset files:")
print(path)


# ============================================================
# 2. FIND ALL CSV FILES
# ============================================================

print("\n=== STEP 2: FINDING DATA FILES ===")

csv_files = glob.glob(
    os.path.join(path, "**", "*.csv"),
    recursive=True
)

print(f"\nFound {len(csv_files)} CSV files:\n")

for file in csv_files:
    size_mb = os.path.getsize(file) / (1024 * 1024)
    print(
        f"{os.path.basename(file):50s} "
        f"{size_mb:8.2f} MB"
    )


# ============================================================
# 3. DISPLAY DATASET STRUCTURE
# ============================================================

print("\n=== STEP 3: INSPECTING DATASET ===")

for file in csv_files:

    print("\n" + "-" * 80)
    print(os.path.basename(file))
    print("-" * 80)

    try:

        temp = pd.read_csv(
            file,
            nrows=3
        )

        print(
            "Columns:",
            temp.columns.tolist()
        )

        print(
            "Sample shape:",
            temp.shape
        )

    except Exception as e:

        print(
            "Could not inspect:",
            e
        )


# ============================================================
# 4. IDENTIFY IMPORTANT FILES
# ============================================================

print("\n=== STEP 4: IDENTIFYING BDG2 DATA COMPONENTS ===")

metadata_files = []
weather_files = []
meter_files = []

for file in csv_files:

    name = os.path.basename(file).lower()
    full_name = file.lower()

    if "metadata" in name:
        metadata_files.append(file)

    elif "weather" in name:
        weather_files.append(file)

    elif (
        "meter" in name
        or "electric" in name
        or "energy" in name
        or "/meters/" in full_name
    ):
        meter_files.append(file)


print("\nMetadata files:")
for f in metadata_files:
    print(f)

print("\nWeather files:")
for f in weather_files:
    print(f)

print("\nMeter/Energy files:")
for f in meter_files:
    print(f)


# ============================================================
# 5. FIND THE MAIN METER FILE
# ============================================================

print("\n=== STEP 5: FINDING MAIN ENERGY DATA ===")

meter_candidates = []

for file in csv_files:

    try:

        temp = pd.read_csv(
            file,
            nrows=2
        )

        columns_lower = [
            str(c).lower()
            for c in temp.columns
        ]

        score = 0

        if "timestamp" in columns_lower:
            score += 10

        if "meter_id" in columns_lower:
            score += 10

        if "building_id" in columns_lower:
            score += 10

        if "meter" in columns_lower:
            score += 8

        if "energy" in " ".join(columns_lower):
            score += 5

        if "electricity" in " ".join(columns_lower):
            score += 5

        if "weather" in file.lower():
            score -= 20

        if "metadata" in file.lower():
            score -= 20

        meter_candidates.append(
            (score, file)
        )

    except:
        pass


meter_candidates.sort(
    reverse=True
)

main_meter_file = meter_candidates[0][1]

print(
    "\nSelected meter file:"
)

print(main_meter_file)


# ============================================================
# 6. LOAD METER DATA
# ============================================================

print("\n=== STEP 6: LOADING REAL ENERGY DATA ===")

meter_df = pd.read_csv(
    main_meter_file
)

print(
    "\nMeter data shape:",
    meter_df.shape
)

print(
    "\nColumns:"
)

print(
    meter_df.columns.tolist()
)

display(
    meter_df.head()
)


# ============================================================
# 7. DETERMINE DATA FORMAT
# ============================================================

print("\n=== STEP 7: ANALYZING METER FORMAT ===")

print(
    "Number of rows:",
    len(meter_df)
)

print(
    "Number of columns:",
    len(meter_df.columns)
)


# ============================================================
# 8. FIND TIMESTAMP COLUMN
# ============================================================

timestamp_candidates = [
    "timestamp",
    "Timestamp",
    "datetime",
    "Datetime",
    "date",
    "Date"
]

timestamp_col = None

for col in timestamp_candidates:

    if col in meter_df.columns:

        timestamp_col = col
        break


if timestamp_col is None:

    for col in meter_df.columns:

        converted = pd.to_datetime(
            meter_df[col],
            errors="coerce"
        )

        if converted.notna().mean() > 0.8:

            timestamp_col = col
            break


if timestamp_col is None:

    raise ValueError(
        "Timestamp column could not be identified."
    )


print(
    "\nTimestamp column:",
    timestamp_col
)

meter_df[timestamp_col] = pd.to_datetime(
    meter_df[timestamp_col],
    errors="coerce"
)

meter_df = meter_df.dropna(
    subset=[timestamp_col]
)


# ============================================================
# 9. DETECT WIDE OR LONG FORMAT
# ============================================================

print("\n=== STEP 8: DETECTING METER FORMAT ===")

numeric_columns = meter_df.select_dtypes(
    include=np.number
).columns.tolist()

print(
    "Numerical columns:",
    len(numeric_columns)
)

print(
    numeric_columns[:20]
)


# BDG2 meter data is commonly represented with
# timestamp rows and meter/building columns.
#
# We identify whether the file contains many
# individual meter columns.

id_columns = [
    timestamp_col,
    "building_id",
    "meter_id",
    "meter"
]

existing_ids = [
    c for c in id_columns
    if c in meter_df.columns
]

value_columns = [
    c for c in numeric_columns
    if c not in existing_ids
]


# ============================================================
# 10. CONVERT TO OBSERVATION FORMAT
# ============================================================

print("\n=== STEP 9: CREATING REAL OBSERVATIONS ===")

if len(value_columns) > 20:

    print(
        "Detected wide-format meter data."
    )

    energy_df = meter_df[
        [timestamp_col] + value_columns
    ].melt(
        id_vars=[timestamp_col],
        var_name="building_id",
        value_name="energy_consumption"
    )

else:

    print(
        "Detected long-format meter data."
    )

    energy_col = None

    possible_names = [
        "energy_consumption",
        "consumption",
        "energy",
        "electricity",
        "value",
        "meter_reading"
    ]

    for name in possible_names:

        matches = [
            c for c in meter_df.columns
            if name in c.lower()
        ]

        if matches:

            energy_col = matches[0]
            break

    if energy_col is None:

        if len(value_columns) > 0:
            energy_col = value_columns[0]

        else:
            raise ValueError(
                "Could not identify energy column."
            )

    energy_df = meter_df[
        [timestamp_col, energy_col]
    ].copy()

    energy_df = energy_df.rename(
        columns={
            energy_col:
            "energy_consumption"
        }
    )

    if "building_id" in meter_df.columns:

        energy_df["building_id"] = (
            meter_df["building_id"]
            .values
        )


print(
    "\nObservation dataset:",
    energy_df.shape
)


# ============================================================
# 11. CLEAN ENERGY DATA
# ============================================================

print("\n=== STEP 10: CLEANING ENERGY DATA ===")

energy_df[
    "energy_consumption"
] = pd.to_numeric(
    energy_df["energy_consumption"],
    errors="coerce"
)

energy_df = energy_df.dropna(
    subset=[
        "energy_consumption"
    ]
)

energy_df = energy_df[
    np.isfinite(
        energy_df["energy_consumption"]
    )
]

energy_df = energy_df[
    energy_df["energy_consumption"] >= 0
]

print(
    "Clean observations:",
    len(energy_df)
)


# ============================================================
# 12. CREATE TIME FEATURES
# ============================================================

print("\n=== STEP 11: CREATING TIME FEATURES ===")

energy_df["hour"] = (
    energy_df[timestamp_col].dt.hour
)

energy_df["day"] = (
    energy_df[timestamp_col].dt.day
)

energy_df["month"] = (
    energy_df[timestamp_col].dt.month
)

energy_df["year"] = (
    energy_df[timestamp_col].dt.year
)

energy_df["day_of_week"] = (
    energy_df[timestamp_col]
    .dt.dayofweek
)

energy_df["day_of_year"] = (
    energy_df[timestamp_col]
    .dt.dayofyear
)

energy_df["week_of_year"] = (
    energy_df[timestamp_col]
    .dt.isocalendar()
    .week
    .astype(int)
)

energy_df["quarter"] = (
    energy_df[timestamp_col].dt.quarter
)

energy_df["is_weekend"] = (
    energy_df["day_of_week"] >= 5
).astype(int)

energy_df["is_business_hour"] = (
    (energy_df["hour"] >= 8)
    &
    (energy_df["hour"] <= 18)
    &
    (energy_df["is_weekend"] == 0)
).astype(int)

energy_df["is_night"] = (
    (energy_df["hour"] < 6)
    |
    (energy_df["hour"] >= 22)
).astype(int)


# ============================================================
# 13. CYCLICAL TIME FEATURES
# ============================================================

print("\n=== STEP 12: CREATING CYCLICAL FEATURES ===")

energy_df["hour_sin"] = np.sin(
    2 * np.pi *
    energy_df["hour"] / 24
)

energy_df["hour_cos"] = np.cos(
    2 * np.pi *
    energy_df["hour"] / 24
)

energy_df["month_sin"] = np.sin(
    2 * np.pi *
    energy_df["month"] / 12
)

energy_df["month_cos"] = np.cos(
    2 * np.pi *
    energy_df["month"] / 12
)

energy_df["weekday_sin"] = np.sin(
    2 * np.pi *
    energy_df["day_of_week"] / 7
)

energy_df["weekday_cos"] = np.cos(
    2 * np.pi *
    energy_df["day_of_week"] / 7
)

energy_df["dayofyear_sin"] = np.sin(
    2 * np.pi *
    energy_df["day_of_year"] / 365
)

energy_df["dayofyear_cos"] = np.cos(
    2 * np.pi *
    energy_df["day_of_year"] / 365
)


# ============================================================
# 14. ENERGY LAG FEATURES
# ============================================================

print("\n=== STEP 13: CREATING ENERGY HISTORY FEATURES ===")

if "building_id" in energy_df.columns:

    energy_df = energy_df.sort_values(
        [
            "building_id",
            timestamp_col
        ]
    )

    grouped = energy_df.groupby(
        "building_id"
    )

    energy_df["energy_lag_1"] = (
        grouped[
            "energy_consumption"
        ].shift(1)
    )

    energy_df["energy_lag_24"] = (
        grouped[
            "energy_consumption"
        ].shift(24)
    )

    energy_df["energy_lag_168"] = (
        grouped[
            "energy_consumption"
        ].shift(168)
    )

    energy_df["rolling_mean_3"] = (
        grouped[
            "energy_consumption"
        ]
        .transform(
            lambda x:
            x.rolling(
                3,
                min_periods=1
            ).mean()
        )
    )

    energy_df["rolling_mean_24"] = (
        grouped[
            "energy_consumption"
        ]
        .transform(
            lambda x:
            x.rolling(
                24,
                min_periods=1
            ).mean()
        )
    )

    energy_df["rolling_std_24"] = (
        grouped[
            "energy_consumption"
        ]
        .transform(
            lambda x:
            x.rolling(
                24,
                min_periods=2
            ).std()
        )
    )

else:

    energy_df = energy_df.sort_values(
        timestamp_col
    )

    energy_df["energy_lag_1"] = (
        energy_df[
            "energy_consumption"
        ].shift(1)
    )

    energy_df["energy_lag_24"] = (
        energy_df[
            "energy_consumption"
        ].shift(24)
    )

    energy_df["energy_lag_168"] = (
        energy_df[
            "energy_consumption"
        ].shift(168)
    )

    energy_df["rolling_mean_3"] = (
        energy_df[
            "energy_consumption"
        ]
        .rolling(
            3,
            min_periods=1
        )
        .mean()
    )

    energy_df["rolling_mean_24"] = (
        energy_df[
            "energy_consumption"
        ]
        .rolling(
            24,
            min_periods=1
        )
        .mean()
    )

    energy_df["rolling_std_24"] = (
        energy_df[
            "energy_consumption"
        ]
        .rolling(
            24,
            min_periods=2
        )
        .std()
    )


# ============================================================
# 15. ADD ENERGY STATISTICAL FEATURES
# ============================================================

energy_df["energy_log"] = np.log1p(
    energy_df[
        "energy_consumption"
    ]
)

energy_df["energy_sqrt"] = np.sqrt(
    energy_df[
        "energy_consumption"
    ]
)

energy_df["energy_squared"] = (
    energy_df[
        "energy_consumption"
    ] ** 2
)


# ============================================================
# 16. CREATE TARGET
# ============================================================

print("\n=== STEP 14: CREATING TARGET FROM REAL ENERGY ===")

q33 = energy_df[
    "energy_consumption"
].quantile(0.33)

q66 = energy_df[
    "energy_consumption"
].quantile(0.66)

print(
    f"33rd percentile: {q33:.4f}"
)

print(
    f"66th percentile: {q66:.4f}"
)


def energy_class(value):

    if value <= q33:
        return 0

    elif value <= q66:
        return 1

    return 2


energy_df["target"] = (
    energy_df[
        "energy_consumption"
    ]
    .apply(energy_class)
)


print(
    "\nTarget distribution:"
)

print(
    energy_df[
        "target"
    ].value_counts()
    .sort_index()
)


# ============================================================
# 17. REMOVE DIRECT TARGET LEAKAGE
# ============================================================

print("\n=== STEP 15: REMOVING TARGET LEAKAGE ===")

# Current energy consumption directly determines target,
# so it MUST NOT be given to the model.

drop_columns = [
    "target",
    "energy_consumption",
    "energy_log",
    "energy_sqrt",
    "energy_squared"
]

model_df = energy_df.drop(
    columns=[
        c for c in drop_columns
        if c in energy_df.columns
    ]
)

y = energy_df["target"]


# ============================================================
# 18. REMOVE ROWS WITH INSUFFICIENT HISTORY
# ============================================================

history_columns = [
    "energy_lag_1",
    "energy_lag_24",
    "energy_lag_168"
]

existing_history = [
    c for c in history_columns
    if c in model_df.columns
]

model_df = model_df.dropna(
    subset=existing_history
)

y = y.loc[
    model_df.index
]


# ============================================================
# 19. SELECT 100,000 REAL ROWS
# ============================================================

print("\n=== STEP 16: SELECTING 100,000 REAL ROWS ===")

if len(model_df) < 100000:

    raise ValueError(
        f"Only {len(model_df):,} usable rows available."
    )

selected = np.random.choice(
    model_df.index,
    size=100000,
    replace=False
)

model_df = model_df.loc[
    selected
].reset_index(drop=True)

y = y.loc[
    selected
].reset_index(drop=True)

print(
    "Final raw observations:",
    len(model_df)
)


# ============================================================
# 20. LOAD METADATA
# ============================================================

print("\n=== STEP 17: ADDING BUILDING METADATA ===")

if len(metadata_files) > 0:

    metadata_file = metadata_files[0]

    print(
        "Metadata file:",
        metadata_file
    )

    metadata = pd.read_csv(
        metadata_file
    )

    print(
        "Metadata shape:",
        metadata.shape
    )

    print(
        "Metadata columns:"
    )

    print(
        metadata.columns.tolist()
    )

    if (
        "building_id" in model_df.columns
        and
        "building_id" in metadata.columns
    ):

        model_df = model_df.merge(
            metadata,
            on="building_id",
            how="left",
            suffixes=(
                "",
                "_metadata"
            )
        )

        print(
            "Metadata successfully merged."
        )

    else:

        print(
            "Building ID not available "
            "for metadata merge."
        )


# ============================================================
# 21. FIND WEATHER DATA
# ============================================================

print("\n=== STEP 18: ADDING WEATHER FEATURES ===")

if len(weather_files) > 0:

    weather_file = weather_files[0]

    print(
        "Weather file:",
        weather_file
    )

    weather = pd.read_csv(
        weather_file
    )

    print(
        "Weather shape:",
        weather.shape
    )

    print(
        "Weather columns:"
    )

    print(
        weather.columns.tolist()
    )

else:

    weather = None

    print(
        "No separate weather CSV detected."
    )


# ============================================================
# 22. PREPARE MODEL FEATURES
# ============================================================

print("\n=== STEP 19: PREPARING FEATURES ===")

if timestamp_col in model_df.columns:

    model_df = model_df.drop(
        columns=[timestamp_col]
    )


# Remove completely empty columns
model_df = model_df.dropna(
    axis=1,
    how="all"
)


# Remove columns with excessive missing values
missing_ratio = (
    model_df.isna().mean()
)

keep_columns = missing_ratio[
    missing_ratio < 0.70
].index

model_df = model_df[
    keep_columns
]


# ============================================================
# 23. DETERMINE NUMERICAL AND CATEGORICAL FEATURES
# ============================================================

num_cols = model_df.select_dtypes(
    include=np.number
).columns.tolist()

cat_cols = model_df.select_dtypes(
    include=[
        "object",
        "category",
        "bool"
    ]
).columns.tolist()


print(
    "\nNumerical features:",
    len(num_cols)
)

print(
    "Categorical features:",
    len(cat_cols)
)

print(
    "Total raw features:",
    len(num_cols) +
    len(cat_cols)
)


# ============================================================
# 24. MAKE SURE WE HAVE 35+ REAL FEATURES
# ============================================================

if len(num_cols) + len(cat_cols) < 35:

    print(
        "\nWARNING:"
    )

    print(
        "The current data structure contains fewer "
        "than 35 usable raw features."
    )

    print(
        "Raw features:",
        len(num_cols) +
        len(cat_cols)
    )

    print(
        "We will NOT generate fake columns."
    )

else:

    print(
        "\n✔ Requirement satisfied:"
    )

    print(
        f"✔ {len(num_cols) + len(cat_cols)} "
        "real/derived features available."
    )


# ============================================================
# 25. TRAIN TEST SPLIT
# ============================================================

print("\n=== STEP 20: TRAIN / TEST SPLIT ===")

X_train_raw, X_test_raw, y_train, y_test = (
    train_test_split(
        model_df,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )
)

print(
    "Training rows:",
    len(X_train_raw)
)

print(
    "Testing rows:",
    len(X_test_raw)
)


# ============================================================
# 26. NUMERICAL PREPROCESSING
# ============================================================

print("\n=== STEP 21: NUMERICAL PREPROCESSING ===")

if len(num_cols) > 0:

    scaler = StandardScaler()

    train_num = (
        X_train_raw[num_cols]
        .fillna(
            X_train_raw[
                num_cols
            ].median()
        )
    )

    test_num = (
        X_test_raw[num_cols]
        .fillna(
            X_train_raw[
                num_cols
            ].median()
        )
    )

    X_train_num = (
        scaler.fit_transform(
            train_num
        )
    )

    X_test_num = (
        scaler.transform(
            test_num
        )
    )

else:

    X_train_num = np.empty(
        (
            len(X_train_raw),
            0
        )
    )

    X_test_num = np.empty(
        (
            len(X_test_raw),
            0
        )
    )


# ============================================================
# 27. CATEGORICAL PREPROCESSING
# ============================================================

print("\n=== STEP 22: CATEGORICAL PREPROCESSING ===")

if len(cat_cols) > 0:

    train_cat = (
        X_train_raw[
            cat_cols
        ]
        .fillna("Unknown")
        .astype(str)
    )

    test_cat = (
        X_test_raw[
            cat_cols
        ]
        .fillna("Unknown")
        .astype(str)
    )

    try:

        encoder = OneHotEncoder(
            sparse_output=False,
            handle_unknown="ignore"
        )

    except TypeError:

        encoder = OneHotEncoder(
            sparse=False,
            handle_unknown="ignore"
        )

    X_train_cat = (
        encoder.fit_transform(
            train_cat
        )
    )

    X_test_cat = (
        encoder.transform(
            test_cat
        )
    )

    encoded_names = list(
        encoder.get_feature_names_out(
            cat_cols
        )
    )

else:

    X_train_cat = np.empty(
        (
            len(X_train_raw),
            0
        )
    )

    X_test_cat = np.empty(
        (
            len(X_test_raw),
            0
        )
    )

    encoded_names = []


# ============================================================
# 28. COMBINE FEATURES
# ============================================================

X_train = np.hstack(
    [
        X_train_num,
        X_train_cat
    ]
)

X_test = np.hstack(
    [
        X_test_num,
        X_test_cat
    ]
)

processed_feature_names = (
    num_cols +
    encoded_names
)

print(
    "\nFinal processed train shape:",
    X_train.shape
)

print(
    "Final processed test shape:",
    X_test.shape
)

print(
    "Final processed feature count:",
    len(processed_feature_names)
)


# ============================================================
# 29. SAVE FINAL 100K DATASET
# ============================================================

print("\n=== STEP 23: SAVING FINAL DATASET ===")

full_processed = np.vstack(
    [
        X_train,
        X_test
    ]
)

full_target = np.concatenate(
    [
        y_train.to_numpy(),
        y_test.to_numpy()
    ]
)

final_dataset = pd.DataFrame(
    full_processed,
    columns=processed_feature_names
)

final_dataset["target"] = (
    full_target
)

final_dataset.to_csv(
    "/content/BDG2_final_100k.csv",
    index=False
)

print(
    "\n✔ Saved:"
)

print(
    "/content/BDG2_final_100k.csv"
)

print(
    "Shape:",
    final_dataset.shape
)


# ============================================================
# 30. TRAIN LIGHTGBM
# ============================================================

print("\n" + "=" * 80)
print("STEP 24: TRAINING LIGHTGBM")
print("=" * 80)

start = time.time()

lgb_model = lgb.LGBMClassifier(
    objective="multiclass",
    num_class=3,
    n_estimators=150,
    learning_rate=0.05,
    num_leaves=31,
    random_state=42,
    n_jobs=-1,
    deterministic=True,
    force_col_wise=True,
    verbosity=-1
)

lgb_model.fit(
    X_train,
    y_train
)

lgb_time = (
    time.time() - start
)

y_pred_lgb = (
    lgb_model.predict(
        X_test
    )
)


# ============================================================
# 31. TRAIN RANDOM FOREST
# ============================================================

print("\n" + "=" * 80)
print("STEP 25: TRAINING RANDOM FOREST")
print("=" * 80)

start = time.time()

rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(
    X_train,
    y_train
)

rf_time = (
    time.time() - start
)

y_pred_rf = (
    rf_model.predict(
        X_test
    )
)


# ============================================================
# 32. MODEL EVALUATION
# ============================================================

print("\n" + "=" * 80)
print("STEP 26: MODEL COMPARISON")
print("=" * 80)


def evaluate(
    name,
    y_true,
    y_pred,
    training_time
):

    return {
        "Model": name,

        "Accuracy": accuracy_score(
            y_true,
            y_pred
        ),

        "Precision": precision_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        ),

        "Recall": recall_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        ),

        "Macro F1": f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        ),

        "Weighted F1": f1_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        ),

        "Training Time (s)": training_time
    }


results = pd.DataFrame(
    [
        evaluate(
            "LightGBM",
            y_test,
            y_pred_lgb,
            lgb_time
        ),

        evaluate(
            "Random Forest",
            y_test,
            y_pred_rf,
            rf_time
        )
    ]
)

display(
    results.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "Macro F1": "{:.4f}",
        "Weighted F1": "{:.4f}",
        "Training Time (s)": "{:.3f}"
    })
)


# ============================================================
# 33. CLASSIFICATION REPORTS
# ============================================================

print(
    "\n=== LIGHTGBM CLASSIFICATION REPORT ==="
)

print(
    classification_report(
        y_test,
        y_pred_lgb,
        target_names=[
            "Low",
            "Medium",
            "High"
        ],
        digits=4,
        zero_division=0
    )
)


print(
    "\n=== RANDOM FOREST CLASSIFICATION REPORT ==="
)

print(
    classification_report(
        y_test,
        y_pred_rf,
        target_names=[
            "Low",
            "Medium",
            "High"
        ],
        digits=4,
        zero_division=0
    )
)


# ============================================================
# 34. CONFUSION MATRICES
# ============================================================

cm_lgb = confusion_matrix(
    y_test,
    y_pred_lgb
)

cm_rf = confusion_matrix(
    y_test,
    y_pred_rf
)

fig, axes = plt.subplots(
    1,
    2,
    figsize=(14, 5)
)

sns.heatmap(
    cm_lgb,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Low",
        "Medium",
        "High"
    ],
    yticklabels=[
        "Low",
        "Medium",
        "High"
    ],
    ax=axes[0]
)

axes[0].set_title(
    "LightGBM Confusion Matrix"
)

axes[0].set_xlabel(
    "Predicted"
)

axes[0].set_ylabel(
    "Actual"
)


sns.heatmap(
    cm_rf,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=[
        "Low",
        "Medium",
        "High"
    ],
    yticklabels=[
        "Low",
        "Medium",
        "High"
    ],
    ax=axes[1]
)

axes[1].set_title(
    "Random Forest Confusion Matrix"
)

axes[1].set_xlabel(
    "Predicted"
)

axes[1].set_ylabel(
    "Actual"
)

plt.tight_layout()
plt.show()


# ============================================================
# 35. FEATURE IMPORTANCE
# ============================================================

print(
    "\n=== STEP 27: LIGHTGBM FEATURE IMPORTANCE ==="
)

importance_df = pd.DataFrame({
    "Feature":
        processed_feature_names,

    "Importance":
        lgb_model.feature_importances_
})

importance_df = (
    importance_df
    .sort_values(
        "Importance",
        ascending=False
    )
    .head(20)
)

display(
    importance_df
)

plt.figure(
    figsize=(10, 7)
)

plt.barh(
    importance_df[
        "Feature"
    ][::-1],
    importance_df[
        "Importance"
    ][::-1]
)

plt.xlabel(
    "Importance"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Top 20 LightGBM Features"
)

plt.tight_layout()
plt.show()


# ============================================================
# 36. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("FINAL SUMMARY")
print("=" * 80)

print(
    f"Real BDG2 observations used: "
    f"{len(model_df):,}"
)

print(
    f"Required observations: 100,000"
)

print(
    f"Raw features: "
    f"{len(num_cols) + len(cat_cols)}"
)

print(
    f"Processed features: "
    f"{len(processed_feature_names)}"
)

print(
    f"Training rows: "
    f"{len(X_train):,}"
)

print(
    f"Testing rows: "
    f"{len(X_test):,}"
)

print(
    f"\nLightGBM Accuracy: "
    f"{accuracy_score(y_test, y_pred_lgb):.4f}"
)

print(
    f"Random Forest Accuracy: "
    f"{accuracy_score(y_test, y_pred_rf):.4f}"
)

print(
    "\nFinal dataset:"
)

print(
    "/content/BDG2_final_100k.csv"
)

print(
    "\n✔ EXPERIMENT COMPLETED"
)
```

```python
# ============================================================
# HYPERPARAMETER TUNING
# ============================================================

from sklearn.model_selection import RandomizedSearchCV

print("=" * 80)
print("HYPERPARAMETER TUNING")
print("=" * 80)


# ============================================================
# LIGHTGBM PARAMETER SEARCH
# ============================================================

print("\n=== TUNING LIGHTGBM ===")

lgb_param_grid = {
    "n_estimators": [
        100,
        150,
        200,
        250
    ],

    "learning_rate": [
        0.01,
        0.03,
        0.05,
        0.08,
        0.10
    ],

    "num_leaves": [
        15,
        20,
        31,
        40,
        50
    ],

    "max_depth": [
        -1,
        5,
        8,
        10,
        15
    ],

    "min_child_samples": [
        10,
        20,
        30,
        50
    ],

    "subsample": [
        0.8,
        0.9,
        1.0
    ],

    "colsample_bytree": [
        0.8,
        0.9,
        1.0
    ]
}


lgb_base = lgb.LGBMClassifier(
    objective="multiclass",
    num_class=3,
    random_state=42,
    n_jobs=-1,
    deterministic=True,
    force_col_wise=True,
    verbosity=-1
)


lgb_search = RandomizedSearchCV(
    estimator=lgb_base,
    param_distributions=lgb_param_grid,
    n_iter=20,
    scoring="accuracy",
    cv=3,
    verbose=1,
    random_state=42,
    n_jobs=1,
    return_train_score=False
)


start_time = time.time()

lgb_search.fit(
    X_train,
    y_train
)

lgb_tuning_time = (
    time.time() - start_time
)


print("\nBest LightGBM Parameters:")

print(
    lgb_search.best_params_
)

print(
    "\nBest LightGBM CV Accuracy:"
)

print(
    f"{lgb_search.best_score_:.4f}"
)


# ============================================================
# RANDOM FOREST PARAMETER SEARCH
# ============================================================

print("\n=== TUNING RANDOM FOREST ===")

rf_param_grid = {
    "n_estimators": [
        100,
        150,
        200,
        300
    ],

    "max_depth": [
        5,
        10,
        15,
        20,
        25,
        None
    ],

    "min_samples_split": [
        2,
        5,
        10,
        20
    ],

    "min_samples_leaf": [
        1,
        2,
        4,
        8
    ],

    "max_features": [
        "sqrt",
        "log2",
        None
    ]
}


rf_base = RandomForestClassifier(
    random_state=42,
    n_jobs=-1
)


rf_search = RandomizedSearchCV(
    estimator=rf_base,
    param_distributions=rf_param_grid,
    n_iter=20,
    scoring="accuracy",
    cv=3,
    verbose=1,
    random_state=42,
    n_jobs=1,
    return_train_score=False
)


start_time = time.time()

rf_search.fit(
    X_train,
    y_train
)

rf_tuning_time = (
    time.time() - start_time
)


print("\nBest Random Forest Parameters:")

print(
    rf_search.best_params_
)

print(
    "\nBest Random Forest CV Accuracy:"
)

print(
    f"{rf_search.best_score_:.4f}"
)


# ============================================================
# EVALUATE TUNED MODELS ON TEST SET
# ============================================================

print("\n" + "=" * 80)
print("EVALUATING TUNED MODELS")
print("=" * 80)


best_lgb = lgb_search.best_estimator_

best_rf = rf_search.best_estimator_


tuned_lgb_pred = (
    best_lgb.predict(
        X_test
    )
)

tuned_rf_pred = (
    best_rf.predict(
        X_test
    )
)


tuned_results = pd.DataFrame([
    {
        "Model": "Tuned LightGBM",

        "Accuracy":
            accuracy_score(
                y_test,
                tuned_lgb_pred
            ),

        "Precision":
            precision_score(
                y_test,
                tuned_lgb_pred,
                average="macro",
                zero_division=0
            ),

        "Recall":
            recall_score(
                y_test,
                tuned_lgb_pred,
                average="macro",
                zero_division=0
            ),

        "Macro F1":
            f1_score(
                y_test,
                tuned_lgb_pred,
                average="macro",
                zero_division=0
            )
    },

    {
        "Model": "Tuned Random Forest",

        "Accuracy":
            accuracy_score(
                y_test,
                tuned_rf_pred
            ),

        "Precision":
            precision_score(
                y_test,
                tuned_rf_pred,
                average="macro",
                zero_division=0
            ),

        "Recall":
            recall_score(
                y_test,
                tuned_rf_pred,
                average="macro",
                zero_division=0
            ),

        "Macro F1":
            f1_score(
                y_test,
                tuned_rf_pred,
                average="macro",
                zero_division=0
            )
    }
])


display(
    tuned_results.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "Macro F1": "{:.4f}"
    })
)


# ============================================================
# TUNED CLASSIFICATION REPORTS
# ============================================================

print(
    "\n=== TUNED LIGHTGBM CLASSIFICATION REPORT ==="
)

print(
    classification_report(
        y_test,
        tuned_lgb_pred,
        target_names=[
            "Low",
            "Medium",
            "High"
        ],
        digits=4,
        zero_division=0
    )
)


print(
    "\n=== TUNED RANDOM FOREST CLASSIFICATION REPORT ==="
)

print(
    classification_report(
        y_test,
        tuned_rf_pred,
        target_names=[
            "Low",
            "Medium",
            "High"
        ],
        digits=4,
        zero_division=0
    )
)
```
